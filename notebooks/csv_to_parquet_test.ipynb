## Static Catchment Attributes

```python
countries = ["camelsaus", "camelsbr", "camelsgb", "hysets", "lamah"]

for country in countries:
    data_folder = Path(
        f"/Users/cooper/Desktop/Caravan/CARAVAN_FULL/usr/local/google/home/kratzert/Data/Caravan-Jan25-csv/attributes/{country}"
    )
    output_folder = Path(
        f"/Users/cooper/Desktop/CaravanifyParquet/{country}/post_processed/attributes/{country}"
    )
    output_folder.mkdir(parents=True, exist_ok=True)

    csv_files = list(data_folder.glob("*.csv"))

    print(f"Converting files for {country}...")

    for file in tqdm(csv_files, desc=f"Processing {country}", unit="file"):
        # Replace .csv extension with .parquet
        parquet_file = output_folder / file.with_suffix(".parquet").name

        data = pd.read_csv(file)

        # Save as parquet using pyarrow
        data.to_parquet(parquet_file, index=False, engine="pyarrow")

    print(f"Done converting {country} files to parquet.")

    # Print the difference in disk space
    original_size = sum(file.stat().st_size for file in csv_files)
    parquet_size = sum(
        (output_folder / file.with_suffix(".parquet").name).stat().st_size
        for file in csv_files
    )
    print(f"Original size: {original_size / (1024 * 1024):.2f} MB")
    print(f"Parquet size: {parquet_size / (1024 * 1024):.2f} MB\n")
```
